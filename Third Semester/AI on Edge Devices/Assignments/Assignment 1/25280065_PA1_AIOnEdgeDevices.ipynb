# 4. Depthwise Separable Convolution (Depthwise + Pointwise)

⚬	Output Tensor Dimensions: ‭$(1, 128, 56, 56)$‬‭‬‭‬ ‭‬‭‬ ‭‬‭‬ ‭‬‭‬


⚬	Number of Weights: ‭$Weights_{depthwise} + Weights_{pointwise}$‬‭‬

⚬	Number of Parameters: ‭$Params_{depthwise} + Params_{pointwise}$‬‭‬
⚬	‭$640 + 8,320 =$‬‭‬‭‬‭‬‭‬ ‭$8,960$‬‭‬‭‬

⚬	Number of MACs: ‭$MACs_{depthwise} + MACs_{pointwise}$‬‭‬
⚬	‭$1,806,336 + 25,690,112 =$‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬ ‭$27,496,448$‬‭‬‭‬‭‬‭‬

⚬	Number of FLOPs: ‭$2 \times Total MACs$‬‭‬
⚬	‭$3,612,672 + 51,380,224 =$‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬‭‬ ‭$54,992,896$‬‭‬‭‬‭‬‭‬

Efficiency Gains (Reduction Ratios)


| Metric | Calculation Step | Result |
| :--- | :--- | :--- |
| **Parameter Reduction** | $1 - \frac{\text{Total Params}_{\text{separable}}}{\text{Total Params}_{\text{standard}}} \rightarrow 1 - \frac{8,960}{73,856}$ | $\sim$ **0.8787** (87.87% fewer parameters) |
| **MACs Reduction** | $1 - \frac{\text{Total MACs}_{\text{separable}}}{\text{Total MACs}_{\text{standard}}} \rightarrow 1 - \frac{27,496,448}{231,211,008}$ | $\sim$ **0.8811** (88.11% fewer operations) |


```python
def out_put_tension_diemsion(Batch_size,filter,H,W,C_out,P,S):
    H_out = (H - filter + 2*P)/S + 1
    W_out = (W - filter + 2*P)/S + 1
    return Batch_size,int(C_out),int(H_out),int(W_out)
def number_of_weights(filter,C_in,C_out):
    if(filter==0):
        return C_in*C_out
    if(C_out==0):
        return filter*filter*C_in
    return filter*filter*C_in*C_out
def number_of_parameters(number_of_weights,C_out):
    if(C_out==0):
        return number_of_weights
    return number_of_weights + C_out
def number_of_MACs(number_of_weights,H_in,W_in,Batch_size):
        return number_of_weights*H_in*W_in*Batch_size
```

```python
C_in=64
C_out=128
H=56
W=56
padding=1
Stride=1
filter_size=3
Batch_size=1

print("---------1. Standard 3 ×3 convolution from 64 to 128 channels.---------\n")
print("output tensor dimensions",out_put_tension_diemsion(Batch_size,filter_size,H,W,C_out,padding,Stride))
number_of_weight=number_of_weights(filter_size,C_in,C_out)
print("number of weights",number_of_weight)
number_of_parameters_standard=number_of_parameters(number_of_weight,C_out)
print("number of parameters",number_of_parameters_standard)
number_of_MACs_standard=number_of_MACs(number_of_weight,H,W,Batch_size)
print("number of MACs",number_of_MACs_standard)
print("number of FLOPs",2*number_of_MACs_standard)


print("\n---------2. Depthwise 3 ×3 convolution on 64 channels.---------")
print("output tensor dimensions",out_put_tension_diemsion(Batch_size,filter_size,H,W,C_in,padding,Stride))
number_of_weight=number_of_weights(filter_size,C_in,0)
print("number of weights",number_of_weight)
print("number of parameters",number_of_parameters(number_of_weight,C_in))
number_of_MACs_deptwise=number_of_MACs(number_of_weight,H,W,Batch_size)
print("number of MACs",number_of_MACs_deptwise)
print("number of FLOPs",2*number_of_MACs_deptwise)

print("\n---------3. Pointwise 1 ×1 convolution from 64 to 128 channels.---------")
print("output tensor dimensions",out_put_tension_diemsion(Batch_size,filter_size,H,W,C_out,padding,Stride))
number_of_weight=number_of_weights(0,C_in,C_out)
print("number of weights",number_of_weight)
print("number of parameters",number_of_parameters(number_of_weight,C_out))
print("number of MACs",number_of_MACs(number_of_weight,H,W,Batch_size))
print("number of FLOPs",2*number_of_MACs(number_of_weight,H,W,Batch_size))

print("\n---------4. Depthwise 3 ×3 convolution followed by pointwise 1 ×1 convolution from 64 to 128 channels.---------")
print("output tensor dimensions",out_put_tension_diemsion(Batch_size,filter_size,H,W,C_out,padding,Stride))
number_of_weight_pointwise=number_of_weights(0,C_in,C_out)
number_of_weight_deptwise=number_of_weights(filter_size,C_in,0)
number_of_weight=number_of_weight_pointwise+number_of_weight_deptwise
print("number of weights",number_of_weight)
number_of_parameters_pointwise=number_of_parameters(number_of_weight_pointwise,C_out)
number_of_parameters_deptwise=number_of_parameters(number_of_weight_deptwise,C_in)
number_of_parameters_=number_of_parameters_pointwise+number_of_parameters_deptwise
print("number of parameters",number_of_parameters_)
print("number of MACs",number_of_MACs(number_of_weight,H,W,Batch_size))
print("number of FLOPs",2*number_of_MACs(number_of_weight,H,W,Batch_size))
print("Parameter reduction ratio",1-(number_of_parameters_/number_of_parameters_standard))
print("MACs reduction ratio",(1-(number_of_MACs(number_of_weight,H,W,Batch_size)/number_of_MACs_standard)))
```

```python
def heads_diemsion(embeddings_dimension,number_of_heads):
    return embeddings_dimension/number_of_heads
```

```python
sequence_length_N=512
embeddings_dimension_D=768
number_of_heads_H=12

print("\nThe dimension dh of each attention head",heads_diemsion(embeddings_dimension_D,number_of_heads_H))
print("The dimensions of Q, K, and V.",sequence_length_N,"x",embeddings_dimension_D,"for each of the three matrices")
print("The dimensions of the attention matrix QK⊤for one attention head is .",sequence_length_N,"x",sequence_length_N)
print("4. The total number of parameters in:")
print("• WQ, WK , and WV ,",3*embeddings_dimension_D*embeddings_dimension_D)
print("• WO,",embeddings_dimension_D*embeddings_dimension_D)
print("• the two MLP layers.",embeddings_dimension_D*4*embeddings_dimension_D+4*embeddings_dimension_D*embeddings_dimension_D)
MACs_5=3*sequence_length_N*embeddings_dimension_D*embeddings_dimension_D
print("5. The MACs required for the Q, K, and V projections.",MACs_5)
MACs_6=number_of_heads_H*sequence_length_N*sequence_length_N*heads_diemsion(embeddings_dimension_D,number_of_heads_H)
print("6. The MACs required to compute QK⊤across all attention heads.",MACs_6)
MACs_7=number_of_heads_H*sequence_length_N*sequence_length_N*heads_diemsion(embeddings_dimension_D,number_of_heads_H)
print("7. The MACs required to compute AV across all attention heads.",MACs_7)
MACs_8=sequence_length_N*embeddings_dimension_D*embeddings_dimension_D
print("8. The MACs required for the attention output projection.",MACs_8)
MACs_9=8*sequence_length_N*embeddings_dimension_D*embeddings_dimension_D
print("9. The MACs required for the MLP.",MACs_9)
MACs_total=MACs_5+MACs_6+MACs_7+MACs_8+MACs_9
print("10. The total MACs and FLOPs for the complete Transformer block.",MACs_total," and FLOPS ",2*MACs_total)




```

```python
import torch.nn as nn
import torch
import time
from thop import profile
devices = ["cpu"]
if torch.mps.is_available():
    devices.append("mps")
x = torch.randn(1, 64, 56, 56)
convolutions = {
    "Standard 3x3": nn.Sequential(
        nn.Conv2d(64, 128, kernel_size=3, padding=1, bias=False)
    ),
    "Depthwise 3x3": nn.Sequential(
        nn.Conv2d(64, 64, kernel_size=3, padding=1, groups=64, bias=False)
    ),
    "Pointwise 1x1": nn.Sequential(
        nn.Conv2d(64, 128, kernel_size=1, bias=False)
    ),
    "Depthwise Separable": nn.Sequential(
        nn.Conv2d(64, 64, kernel_size=3, padding=1, groups=64, bias=False),
        nn.Conv2d(64, 128, kernel_size=1, bias=False)
    )
}
```

```python
for device_name in devices:
    print(f"\n--- Profiling on {device_name.upper()} ---")
    device = torch.device(device_name)
    curr_x = x.to(device)
    for name, model in convolutions.items():
        model = model.to(device)
        macs, params = profile(model, inputs=(curr_x,), verbose=False)
        flops = macs * 2
        # Warm-up
        for _ in range(10): model(curr_x)
        if device_name == "mps": torch.mps.synchronize()
        # Measure Latency
        start = time.perf_counter()
        for _ in range(100): model(curr_x)
        if device_name == "mps": torch.mps.synchronize()
        latency_ms = ((time.perf_counter() - start) / 100) * 1000
        # Measure Peak Memory
        mem_mb = 0.0
        if device_name == "mps":
            try:
                mem_mb = torch.mps.current_allocated_memory() / (1024 ** 2)
            except AttributeError:
                pass
        print(f"{name:<16} | Params: {int(params):<6} | MACs: {int(macs):<10} | FLOPs: {int(flops):<10} | Latency: {latency_ms:.3f}ms | Mem: {mem_mb:.2f}MB")
```

## Part (b): Transformer Profiling

```python
device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")

class TransformerBlock(nn.Module):
    def __init__(self, d=768, h=12, dff=3072):
        super().__init__()
        self.h, self.d_k = h, d // h
        self.q_proj, self.k_proj, self.v_proj, self.out_proj = [nn.Linear(d, d) for _ in range(4)]
        self.mlp = nn.Sequential(nn.Linear(d, dff), nn.GELU(), nn.Linear(dff, d))
        
    def forward(self, x):
        B, N, D = x.shape
        p = lambda l: l(x).view(B, N, self.h, self.d_k).transpose(1, 2)
        q, k, v = p(self.q_proj), p(self.k_proj), p(self.v_proj)
        ctx = torch.matmul(torch.softmax(torch.matmul(q, k.transpose(-2, -1)) / (self.d_k**0.5), dim=-1), v).transpose(1, 2).reshape(B, N, D)
        return (x := x + self.out_proj(ctx)) + self.mlp(x)

def sync(d):
    if d.type == "cuda": torch.cuda.synchronize()
    elif d.type == "mps": torch.mps.synchronize()

model = TransformerBlock().to(device).to(torch.float16).eval()
D = 768  # embedding dim, used for the analytic attention-MACs correction below

print(f"{'N':<6} | {'Params':<10} | {'MACs':<12} | {'FLOPs':<12} | {'Latency':<9} | {'Memory'}")
print("-" * 70)

for N in [128, 256, 512, 1024]:
    x = torch.randn(1, N, 768, device=device, dtype=torch.float16)
    macs_thop, params = profile(model, inputs=(x,), verbose=False)

    attn_macs = 2 * (N ** 2) * D
    macs = macs_thop + attn_macs

    with torch.no_grad():
        for _ in range(10): model(x)
        sync(device)
        start = time.perf_counter()
        for _ in range(100): model(x)
        sync(device)
        
    latency = ((time.perf_counter() - start) / 100) * 1000
    mem = (torch.cuda.max_memory_allocated(device) / (1024**2)) if device.type == "cuda" else (torch.mps.current_allocated_memory() / (1024**2) if device.type == "mps" else 0.0)
    print(f"{N:<6} | {int(params):<10} | {int(macs):<12} | {int(macs*2):<12} | {latency:6.3f}ms  | {mem:.2f}MB")
```

## Part (c): KV-Cache Profiling

```python
sync = lambda d: torch.mps.synchronize() if d.type == "mps" else None

class KVCacheBlock(nn.Module):
    def __init__(self, d=768, h=12):
        super().__init__()
        self.h, self.d_k = h, d // h
        self.q, self.k, self.v, self.o = [nn.Linear(d, d) for _ in range(4)]
        self.mlp = nn.Sequential(nn.Linear(d, d*4), nn.GELU(), nn.Linear(d*4, d))
        
    def forward(self, x, pk=None, pv=None):
        B, N, D = x.shape
        proj = lambda l: l(x).view(B, N, self.h, self.d_k).transpose(1, 2)
        q, k, v = proj(self.q), proj(self.k), proj(self.v)
        if pk is not None: k, v = torch.cat([pk, k], -2), torch.cat([pv, v], -2)
        ctx = torch.matmul(torch.softmax(torch.matmul(q, k.transpose(-2, -1)) / (self.d_k**0.5), -1), v).transpose(1, 2).reshape(B, N, D)
        return (x := x + self.o(ctx)) + self.mlp(x), k, v

model = KVCacheBlock().to(device).to(torch.float16).eval()

for t in [128, 256, 512, 1024]:
    print(f"\n================ Target Length (t) = {t} ================")
    latencies = {}
    for use_cache in [False, True]:
        mode = "WITH KV-Cache" if use_cache else "WITHOUT KV-Cache"
        print(f"\n--- Mode: {mode} ---")

        sync(device)
        peak_mem = 0.0
        start = time.perf_counter()

        with torch.no_grad():
            pk, pv = None, None
            for i in range(t):
                x = torch.randn(1, (i+1 if not use_cache else 1), 768, device=device, dtype=torch.float16)
                _, pk, pv = model(x, pk if use_cache else None, pv if use_cache else None)
                # track the running max, not just the last reading, so this is a real peak
                if device.type == "mps":
                    peak_mem = max(peak_mem, torch.mps.current_allocated_memory() / (1024**2))

        sync(device)
        total_latency = (time.perf_counter() - start) * 1000  # ms
        avg_latency = total_latency / t
        latencies[mode] = avg_latency
        kv_mem = (2 * 1 * 12 * t * 64 * 2) / (1024**2) if use_cache else 0.0

        print(f"• Total decoding latency: {total_latency:.2f} ms")
        print(f"• Average latency per generated token: {avg_latency:.3f} ms")
        print(f"• Peak GPU memory usage: {peak_mem:.2f} MB")
        print(f"• KV-cache memory usage: {kv_mem:.2f} MB")
        print(f"• Power consumption: N/A (macOS restriction)")
        print(f"• Energy per generated token: N/A (macOS restriction)")

    # Auto-explain whichever mode came out faster, instead of leaving
    # a counter-intuitive result unremarked.
    faster = min(latencies, key=latencies.get)
    if faster == "WITHOUT KV-Cache":
        print(f"\nNote: at t={t}, no-cache was faster per token "
              f"({latencies['WITHOUT KV-Cache']:.3f}ms vs {latencies['WITH KV-Cache']:.3f}ms). "
              f"Likely cause: the torch.cat used to grow the cache every step adds fixed "
              f"overhead that outweighs the recomputation it saves at this sequence length.")
    else:
        print(f"\nNote: at t={t}, KV-cache was faster per token "
              f"({latencies['WITH KV-Cache']:.3f}ms vs {latencies['WITHOUT KV-Cache']:.3f}ms), "
              f"as expected, since it avoids recomputing K/V for all previous tokens.")
```

## Part (d): Roofline Sweep

```python
import math, time
import torch, torch.nn as nn
import pandas as pd
import matplotlib
# matplotlib.use("Agg")
import matplotlib.pyplot as plt

device = torch.device("mps" if torch.backends.mps.is_available() else "cuda" if torch.cuda.is_available() else "cpu")
dtype = torch.float16
bp = torch.finfo(dtype).bits // 8
sync = lambda: torch.mps.synchronize() if device.type == "mps" else (torch.cuda.synchronize() if device.type == "cuda" else None)

def bench(fn):
    for _ in range(10): fn()
    sync()
    times = []
    for _ in range(20):
        sync(); t0 = time.perf_counter(); fn(); sync()
        times.append(time.perf_counter() - t0)
    return sorted(times)[len(times) // 2]

rows = []
def record(sweep, cfg, macs, flops, bytes_mv, lat, i_star):
    ai, thr = flops / bytes_mv, flops / lat
    rows.append(dict(sweep=sweep, config=cfg, MACs=macs, FLOPs=flops, Bytes=bytes_mv,
                      AI=ai, latency_ms=lat * 1e3, GFLOPs=thr / 1e9,
                      regime="compute" if ai > i_star else "memory"))
    return ai, thr

# 1. Characterization
numel = 64 * 1024 * 1024
a, b, c = torch.randn(numel, device=device, dtype=dtype), torch.randn(numel, device=device, dtype=dtype), torch.empty(numel, device=device, dtype=dtype)
beta = (3 * numel * bp) / bench(lambda: torch.add(a, b, out=c))
N_pk = 4096
A, B = torch.randn(N_pk, N_pk, device=device, dtype=dtype), torch.randn(N_pk, N_pk, device=device, dtype=dtype)
p_peak = (2 * N_pk**3) / bench(lambda: torch.matmul(A, B))
i_star = p_peak / beta
print(f"beta={beta/1e9:.2f} GB/s | P_peak={p_peak/1e12:.2f} TFLOP/s | I*={i_star:.2f}\n")

# 2. Matmul sweep
print("--- Matmul ---")
for N in [64, 128, 256, 512, 1024, 2048, 4096, 6144]:
    A, B = torch.randn(N, N, device=device, dtype=dtype), torch.randn(N, N, device=device, dtype=dtype)
    macs = N**3
    lat = bench(lambda: torch.matmul(A, B))
    ai, thr = record("matmul", f"N={N}", macs, 2*macs, bp*3*N**2, lat, i_star)
    print(f"N={N:<5} AI={ai:8.1f} lat={lat*1e3:7.2f}ms {thr/1e12:6.2f} TFLOP/s [{rows[-1]['regime']}]")

# 3. Conv batch sweep
print("\n--- Conv batch ---")
Cin, Cout, K, H, W = 64, 128, 3, 56, 56
conv = nn.Conv2d(Cin, Cout, K, padding=1, bias=False).to(device).to(dtype)
w_bytes = bp * Cin*Cout*K*K
for Bsz in [1, 2, 4, 8, 16, 32, 64, 128]:
    x = torch.randn(Bsz, Cin, H, W, device=device, dtype=dtype)
    macs = Cin*Cout*K*K*H*W*Bsz
    bytes_mv = w_bytes + bp*(x.numel() + Bsz*Cout*H*W)
    lat = bench(lambda: conv(x))
    ai, thr = record("conv_batch", f"B={Bsz}", macs, 2*macs, bytes_mv, lat, i_star)
    print(f"B={Bsz:<4} AI={ai:8.1f} lat={lat*1e3:7.2f}ms {thr/1e9:7.1f} GFLOP/s [{rows[-1]['regime']}]")
i_limit = (2*Cin*Cout*K*K) / (bp*(Cin+Cout))
print(f"lim(B->inf) I(B) = {i_limit:.2f}  |  weight fraction at B=1 = {w_bytes/(w_bytes+bp*(Cin*H*W+Cout*H*W))*100:.1f}%")

# 4. Conv channel/kernel sweep
print("\n--- Conv channel/kernel ---")
Bsz = 8
for Cin, Cout, K in [(16,16,1),(64,64,1),(16,16,3),(64,64,3),(256,256,3),(64,64,5),(128,128,5),(128,128,7),(256,256,7)]:
    conv = nn.Conv2d(Cin, Cout, K, padding=K//2, bias=False).to(device).to(dtype)
    x = torch.randn(Bsz, Cin, H, W, device=device, dtype=dtype)
    macs = Cin*Cout*K*K*H*W*Bsz
    bytes_mv = bp*Cin*Cout*K*K + bp*(x.numel() + Bsz*Cout*H*W)
    lat = bench(lambda: conv(x))
    cfg = f"Cin={Cin},Cout={Cout},K={K}"
    ai, thr = record("conv_channel", cfg, macs, 2*macs, bytes_mv, lat, i_star)
    print(f"{cfg:<22} AI={ai:8.1f} lat={lat*1e3:7.2f}ms {thr/1e9:7.1f} GFLOP/s [{rows[-1]['regime']}]")

# 5. Transformer prefill/decode sweep
class Block(nn.Module):
    def __init__(self, d, h, ff):
        super().__init__()
        self.h, self.hd = h, d // h
        self.qkv = nn.Linear(d, 3*d, bias=False)
        self.proj = nn.Linear(d, d, bias=False)
        self.ff = nn.Sequential(nn.Linear(d, ff, bias=False), nn.GELU(), nn.Linear(ff, d, bias=False))
        self.ln1, self.ln2 = nn.LayerNorm(d), nn.LayerNorm(d)
    def forward(self, x, cache=None):
        Bsz, T, D = x.shape
        q, k, v = self.qkv(self.ln1(x)).chunk(3, dim=-1)
        rs = lambda t: t.view(Bsz, T, self.h, self.hd).transpose(1, 2)
        q, k, v = rs(q), rs(k), rs(v)
        if cache is not None:
            k, v = torch.cat([cache[0], k], 2), torch.cat([cache[1], v], 2)
        att = (q @ k.transpose(-2, -1) / math.sqrt(self.hd)).softmax(-1)
        out = (att @ v).transpose(1, 2).reshape(Bsz, T, D)
        x = x + self.proj(out)
        x = x + self.ff(self.ln2(x))
        return x, (k, v)

print("\n--- Transformer prefill/decode ---")
d, h, ff, L = 512, 8, 2048, 4
blocks = nn.ModuleList([Block(d, h, ff) for _ in range(L)]).to(device).to(dtype)
w_bytes_t = bp * sum(p.numel() for p in blocks.parameters())

def run(x, caches=None):
    new = []
    for i, blk in enumerate(blocks):
        x, kv = blk(x, caches[i] if caches else None)
        new.append(kv)
    return x, new

for N in [32, 64, 128, 256, 512, 1024, 2048, 4096]:
    x = torch.randn(1, N, d, device=device, dtype=dtype)
    p_flops = L * (8*N*d**2 + 4*N*d*ff + 4*N**2*d)
    p_bytes = w_bytes_t + bp*N*d*2
    lat_p = bench(lambda: run(x))
    ai_p, thr_p = record("transformer_prefill", f"N={N}", p_flops//2, p_flops, p_bytes, lat_p, i_star)

    with torch.no_grad():
        _, cache = run(x)
    x1 = torch.randn(1, 1, d, device=device, dtype=dtype)
    d_flops = L * (8*d**2 + 4*d*ff + 4*(N+1)*d)
    d_bytes = w_bytes_t + bp*2*N*d*L
    lat_d = bench(lambda: run(x1, cache))
    ai_d, thr_d = record("transformer_decode", f"N={N}", d_flops//2, d_flops, d_bytes, lat_d, i_star)

    print(f"N={N:<5} prefill AI={ai_p:7.1f} {thr_p/1e9:7.1f} GFLOP/s | decode AI={ai_d:6.2f} {thr_d/1e9:6.1f} GFLOP/s")

# 6. Deliverable table + plot
df = pd.DataFrame(rows)
pd.set_option("display.width", 160)
print("\n=== Full results table ===")
print(df.to_string(index=False, float_format="%.2f"))


best = df.loc[df.GFLOPs.idxmax()]
print(f"\nBest: {best.GFLOPs:.1f} GFLOP/s ({best.sweep} {best.config}) = {best.GFLOPs*1e9/p_peak*100:.1f}% of P_peak")

fig, ax = plt.subplots(figsize=(7, 5.5))
ai_lo, ai_hi = df.AI.min()*0.5, df.AI.max()*2
ax.plot([ai_lo, i_star], [beta*ai_lo/1e9, p_peak/1e9], "k--", lw=1.2)
ax.plot([i_star, ai_hi], [p_peak/1e9, p_peak/1e9], "k-", lw=1.2)
for name, g in df.groupby("sweep"):
    ax.scatter(g.AI, g.GFLOPs, s=30, label=name)
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("Arithmetic Intensity (FLOPs/Byte)"); ax.set_ylabel("Throughput (GFLOP/s)")
ax.legend(fontsize=8); fig.tight_layout()
print("Saved roofline_results.csv and roofline.png")
```

```python
def plot_roofline(sweep_names, title):
    sub = df[df.sweep.isin(sweep_names)]
    fig, ax = plt.subplots(figsize=(6.5, 5))
    ai_lo = min(sub.AI.min(), i_star) * 0.5
    ai_hi = max(sub.AI.max(), i_star) * 2
    ax.plot([ai_lo, i_star], [beta*ai_lo/1e9, p_peak/1e9], "k--", lw=1.2, label="memory roof")
    ax.plot([i_star, ai_hi], [p_peak/1e9, p_peak/1e9], "k-", lw=1.2, label="compute roof")
    ax.axvline(i_star, color="gray", ls=":", lw=1, label=f"I*={i_star:.1f}")
    for name, g in sub.groupby("sweep"):
        ax.scatter(g.AI, g.GFLOPs, s=35, label=name)
    ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlabel("Arithmetic Intensity (FLOPs/Byte)")
    ax.set_ylabel("Throughput (GFLOP/s)")
    ax.set_title(title)
    ax.legend(fontsize=8)
    fig.tight_layout()
    plt.show()

plot_roofline(["matmul"], "Matmul Roofline")
plot_roofline(["conv_batch", "conv_channel"], "Conv2d Roofline")
plot_roofline(["transformer_prefill", "transformer_decode"], "Transformer Prefill/Decode Roofline")
```

Memory vs compute bound. Only two configs stayed memory-bound: matmul N=64 (AI=21.3) and the 1×1 conv with Cin=Cout=16 (AI=8.0)   both below I*=31.37. Everything else crossed into compute-bound.

Transition point. The switch happens right around I*≈31.4 FLOPs/byte, matching Ppeak/β. This lines up with the matmul sweep flipping memory→compute between N=64 and N=128.

Peak reached. Best throughput was 6024.9 GFLOP/s (conv, Cin=Cout=256, K=3)   116% of the measured P_peak. Not a real violation; it just means the single matmul used to measure P_peak underestimated the device's true ceiling, since conv and matmul hit different kernels on MPS.

Why the bend is gradual. Throughput keeps rising well past I* (0.03 → 5.23 TFLOP/s across matmul sizes) instead of snapping to the ceiling, because small workloads don't yet saturate all compute units and launch/overhead costs dominate at small N.

What raised AI. Channel/kernel size was by far the most effective lever   Cin=Cout=16,K=1 to Cin=Cout=256,K=7 raised AI from 8 to 5017 (600×), since FLOPs scale with Cin·Cout·K² while activation bytes barely move. Batch size was weak   128× more batch only took AI from 342 to 384, since FLOPs and bytes both scale linearly with B, so the ratio saturates (matches the analytic limit of 384).
