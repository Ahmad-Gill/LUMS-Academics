## 1. Introduction

### What is Pandas?

**Pandas** (Python Data Analysis Library) is a powerful, open-source data manipulation and analysis library for Python. It provides data structures and functions needed to work with structured data seamlessly.

### Why Use Pandas?

Let's compare working with raw Python vs. Pandas to see why Pandas is essential for data science:

**Raw Python approach:**
```python
# Working with election data using raw Python
data = [
    ['Reagan', 'Republican', 50.7, 1980, 'win'],
    ['Carter', 'Democratic', 41.0, 1980, 'loss'],
    ['Anderson', 'Independent', 6.6, 1980, 'loss']
]

# To find all winners, you'd need to:
winners = []
for row in data:
    if row[4] == 'win':
        winners.append(row)
```

**Pandas approach:**
```python
# Much simpler with Pandas!
winners = elections[elections['Result'] == 'win']
```

### Key Advantages of Pandas:
- **Intuitive**: Easy-to-use data structures (DataFrame, Series)
- **Powerful**: Built-in functions for data cleaning, transformation, and analysis  
- **Efficient**: Optimized for performance with large datasets
- **Flexible**: Handles missing data, different data types, and complex operations
- **Integration**: Works seamlessly with other Python libraries (NumPy, Matplotlib, etc.)

## 4. Data Cleaning

Real-world data is rarely perfect. Let's learn essential techniques for cleaning and preparing data for analysis.


### Handling Missing Values

Let's create some missing data to demonstrate cleaning techniques, since our elections dataset is clean.


````python
# Create a copy of elections data and introduce some missing values for demonstration
elections_dirty = elections.copy()

# Introduce some NaN values
elections_dirty.loc[1, 'Candidate'] = np.nan
elections_dirty.loc[5, '%'] = np.nan
elections_dirty.loc[8, 'Party'] = np.nan

print("Dataset with missing values:")
elections_dirty.head(10)
````

````python
# Check for missing values
print("Missing values per column:")
print(elections_dirty.isnull().sum())

print("\nRows with any missing values:")
print(elections_dirty[elections_dirty.isnull().any(axis=1)])
````

````python
# Method 1: Drop rows with missing values
print("Original shape:", elections_dirty.shape)
elections_dropped = elections_dirty.dropna()
print("After dropping rows with NaN:", elections_dropped.shape)
````

````python
# Method 2: Fill missing values
elections_filled = elections_dirty.copy()

# Fill missing candidate names with "Unknown"
elections_filled['Candidate'].fillna("Unknown", inplace=True)

# Fill missing percentages with the mean
mean_percentage = elections_filled['%'].mean()
elections_filled['%'].fillna(mean_percentage, inplace=True)

# Fill missing parties with "Independent"  
elections_filled['Party'].fillna("Independent", inplace=True)

print("After filling missing values:")
elections_filled[elections_filled.index.isin([1, 5, 8])]
````

## 5. Filtering & Sorting

One of the most powerful features of Pandas is the ability to filter and sort data efficiently.


### Boolean Filtering

Use boolean conditions to filter rows that meet specific criteria.


````python
# Filter for winners only
winners = elections[elections['Result'] == 'win']
print("Election winners:")
print(winners[['Candidate', 'Party', 'Year', '%']])
````

````python
# Filter for high vote percentages (> 50%)
high_votes = elections[elections['%'] > 50]
print("Candidates with >50% of votes:")
print(high_votes[['Candidate', 'Party', 'Year', '%']])
````

````python
# Multiple conditions - use & (and) or | (or)
# Winners who got less than 50% of the vote
narrow_winners = elections[(elections['Result'] == 'win') & (elections['%'] < 50)]
print("Winners with <50% of votes:")
print(narrow_winners[['Candidate', 'Party', 'Year', '%']])
````

````python
# Filter using .isin() for multiple values
parties_of_interest = ['Democratic', 'Republican']
major_parties = elections[elections['Party'].isin(parties_of_interest)]
print("Major party candidates:")
print(major_parties.head())
````

### Sorting Data

Sort your data by one or more columns to find patterns and insights.


````python
# Sort by vote percentage (ascending)
sorted_by_votes = elections.sort_values('%')
print("Sorted by vote percentage (lowest first):")
print(sorted_by_votes[['Candidate', 'Party', 'Year', '%']].head())
````

````python
# Sort by vote percentage (descending)
sorted_desc = elections.sort_values('%', ascending=False)
print("Sorted by vote percentage (highest first):")
print(sorted_desc[['Candidate', 'Party', 'Year', '%']].head())
````

````python
# Sort by multiple columns
multi_sort = elections.sort_values(['Year', '%'], ascending=[True, False])
print("Sorted by Year (ascending) then Vote % (descending):")
print(multi_sort[['Candidate', 'Party', 'Year', '%']].head(8))
````

### Resetting Index

After filtering and sorting, you might want to reset the index to have clean row numbers.

## 6. GroupBy & Aggregations

GroupBy is one of Pandas' most powerful features, allowing you to split data into groups and perform operations on each group.


### Basic GroupBy Operations

Let's group our election data by different categories and see what insights we can find.


````python
# Group by Party and calculate mean vote percentage
party_stats = elections.groupby('Party')['%'].mean()
print("Average vote percentage by party:")
print(party_stats.sort_values(ascending=False))
````

````python
# Count how many times each party appears
party_counts = elections.groupby('Party').size()
print("Number of candidates by party:")
print(party_counts.sort_values(ascending=False))
````

````python
# Multiple aggregations at once
party_summary = elections.groupby('Party')['%'].agg(['mean', 'max', 'min', 'count'])
print("Party vote statistics:")
print(party_summary.round(2))
````

### GroupBy with Multiple Columns

You can group by multiple columns to get more detailed insights.


````python
# Group by Party and Result
party_result = elections.groupby(['Party', 'Result'])['%'].mean()
print("Average vote percentage by party and result:")
print(party_result.round(2))
````

````python
# Convert MultiIndex result to DataFrame for better viewing
party_result_df = party_result.unstack(fill_value=0)
print("Party performance by result (as table):")
print(party_result_df.round(2))
````

### Custom Aggregation Functions

You can also apply custom functions to your grouped data.


````python
# Custom function to find the range (max - min)
def vote_range(series):
    return series.max() - series.min()

# Apply custom function
party_ranges = elections.groupby('Party')['%'].apply(vote_range)
print("Vote percentage range by party:")
print(party_ranges.round(2))
````

## 7. Merging & Joining

Often you'll need to combine data from multiple sources. Pandas provides powerful tools for merging and joining DataFrames.


### Creating Sample DataFrames for Merging

Let's create some additional datasets to demonstrate different types of joins.


````python
# Create additional datasets to demonstrate merging
candidate_info = pd.DataFrame({
    'Candidate': ['Reagan', 'Carter', 'Clinton', 'Bush', 'Obama'],
    'Age_at_Election': [69, 56, 46, 64, 47],
    'Home_State': ['California', 'Georgia', 'Arkansas', 'Texas', 'Illinois']
})

party_info = pd.DataFrame({
    'Party': ['Democratic', 'Republican', 'Independent'],
    'Founded_Year': [1828, 1854, 1971],
    'Color': ['Blue', 'Red', 'Purple']
})

print("Candidate Info:")
print(candidate_info)
print("\nParty Info:")
print(party_info)
````

### Inner Join

An inner join returns only rows that have matching values in both DataFrames.


````python
# Inner join: merge elections with candidate_info
inner_merged = elections.merge(candidate_info, on='Candidate', how='inner')
print("Inner join result (elections + candidate info):")
print(inner_merged[['Candidate', 'Party', 'Year', '%', 'Age_at_Election', 'Home_State']].head())
````

### Left Join

A left join returns all rows from the left DataFrame and matched rows from the right DataFrame.


````python
# Left join: keep all elections, add candidate info where available
left_merged = elections.merge(candidate_info, on='Candidate', how='left')
print("Left join result (all elections + candidate info where available):")
print(f"Original elections: {len(elections)} rows")
print(f"After left join: {len(left_merged)} rows")
print("\nSample with missing candidate info:")
print(left_merged[left_merged['Age_at_Election'].isna()][['Candidate', 'Party', 'Year']].head())
````

````python
# Merge elections with party_info
party_merged = elections.merge(party_info, on='Party', how='left')
print("Elections with party information:")
print(party_merged[['Candidate', 'Party', 'Year', 'Founded_Year', 'Color']].head())
````

### Concatenating DataFrames

Sometimes you want to stack DataFrames vertically (adding rows) or horizontally (adding columns).


````python
# Create two subsets of elections data
elections_1980s = elections[elections['Year'] < 1990]
elections_2000s = elections[elections['Year'] >= 2000]

print("1980s elections:", len(elections_1980s))
print("2000s+ elections:", len(elections_2000s))

# Concatenate vertically (stack rows)
combined_elections = pd.concat([elections_1980s, elections_2000s], axis=0)
print(f"Combined: {len(combined_elections)} rows")
print("\nFirst few rows of combined data:")
print(combined_elections.head())
````

## 8. Advanced Operations

Let's explore some powerful advanced features: apply functions, lambda expressions, and pivot tables.


### Apply Functions

The `apply()` method allows you to apply a function along an axis of a DataFrame or to each element of a Series.


````python
# Apply a function to transform vote percentages to decimals
def to_decimal(percentage):
    return percentage / 100

elections['Vote_Decimal'] = elections['%'].apply(to_decimal)
print("Applied function to convert percentages:")
print(elections[['Candidate', '%', 'Vote_Decimal']].head())
````

````python
# Apply function row-wise (axis=1) to create new columns
def categorize_performance(row):
    if row['%'] >= 50:
        return 'Strong'
    elif row['%'] >= 40:
        return 'Moderate' 
    else:
        return 'Weak'

elections['Performance'] = elections.apply(categorize_performance, axis=1)
print("Applied row-wise function:")
print(elections[['Candidate', '%', 'Performance']].head())
````

### Lambda Functions

Lambda functions provide a concise way to write simple functions inline.


````python
# Using lambda to create candidate name lengths
elections['Name_Length'] = elections['Candidate'].apply(lambda x: len(x))

# Using lambda for more complex operations
elections['Era'] = elections['Year'].apply(lambda year: '1980s' if year < 1990 
                                         else '1990s' if year < 2000
                                         else '2000s' if year < 2010
                                         else '2010s+')

print("Lambda function results:")
print(elections[['Candidate', 'Year', 'Name_Length', 'Era']].head())
````

### String Operations

Pandas provides powerful string manipulation methods through the `.str` accessor.


````python
# String operations on candidate names
elections['Candidate_Upper'] = elections['Candidate'].str.upper()
elections['Candidate_Lower'] = elections['Candidate'].str.lower()
elections['Starts_with_R'] = elections['Candidate'].str.startswith('R')

print("String operations:")
print(elections[['Candidate', 'Candidate_Upper', 'Starts_with_R']].head())
````
